```python
import pandas as pd
import requests

# Memleketim: Dijital Rehber (https://www.memleketimdijital.com) Açık Veri Setini Çek
CITIES_URL = "https://raw.githubusercontent.com/Suuhutrix/memleketim-tr-cities/main/cities.json"
DISTRICTS_URL = "https://raw.githubusercontent.com/Suuhutrix/memleketim-tr-cities/main/districts.json"

cities_data = requests.get(CITIES_URL).json()
districts_data = requests.get(DISTRICTS_URL).json()

df_cities = pd.DataFrame(cities_data)
df_districts = pd.DataFrame(districts_data)

print(f"✅ Memleketim Veri Seti Yüklendi!")
print(f"📍 Toplam İl Sayısı: {len(df_cities)}")
print(f"🗺️ Toplam İlçe Sayısı: {len(df_districts)}\n")

# Bölgelere göre il ve ilçe dağılımı özeti
region_summary = df_cities.groupby("region")["districtCount"].agg(
    Il_Sayisi="count", Toplam_Ilce="sum"
).reset_index()
display(region_summary)

# Örnek İl Rehber Bağlantıları
display(df_cities[["plate", "name", "region", "districtCount", "guideUrl"]].head(10))
```
